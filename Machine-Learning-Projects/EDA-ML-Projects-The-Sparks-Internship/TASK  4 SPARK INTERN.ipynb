```python
## Counting the peoples who are affected from Attacker type
AttackType=data1.pivot_table(columns='AttackType',values='Effected_People',aggfunc='sum')
AttackType = AttackType.T
AttackType['Type'] = AttackType.index
```

```python
import folium
from folium.plugins import MarkerCluster
year=data1[data1['Year']==2014]
mapData=year.loc[:,'City':'longitude']
mapData=mapData.dropna().values.tolist()

map = folium.Map(location = [0, 50], tiles='CartoDB positron', zoom_start=2) 
markerCluster = folium.plugins.MarkerCluster().add_to(map)
for point in range(0, len(mapData)):
    folium.Marker(location=[mapData[point][1],mapData[point][2]],
                  popup = mapData[point][0]).add_to(markerCluster)
map
```

```python
plt.figure(figsize=(15,6))
country_attack=data1.Country.value_counts()[:15].reset_index()
country_attack.columns= ["Country", "Total Attacks"]
px.bar(data_frame= country_attack,x = 'Country',y = 'Total Attacks',color = 'Country',template='plotly_dark')
```

```python
# Check Which Country peoples are Effected more
```

```python
plt.figure(figsize=(15, 8))
cas_count= data1.groupby("Country").Effected_People.sum().to_frame().reset_index().sort_values("Effected_People", ascending=False)[:15]
px.bar(data_frame=cas_count,x = 'Country',y = 'Effected_People',color='Country',template='plotly_dark')
```
